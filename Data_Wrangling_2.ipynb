```python
import pandas as pd
data={'Name':['Sai','Yuvasri','Poovannan','Sabharish','Sangamithra','Rithika','Rubasri'],'Age':[20,18,19,20,18,17,19],'Gender':['M','F','M','M','F','F','F'],'Marks':[90,76,'NaN',74,65,'NaN',71]}
df=pd.DataFrame(data)
print("Original DataFrame:\n",df)

c = 0
total_marks = 0

for ele in df['Marks']:
  if str(ele).isnumeric():
    c += 1
    total_marks += ele

# Calculate average after the loop
if c > 0:
    average_marks = total_marks / c
else:
    average_marks = 0 # Default to 0 if no numeric marks found

# Replace 'NaN' string with the calculated average
df = df.replace(to_replace="NaN", value=average_marks)

print("\nReplacing NaN with average marks:\n",df)
```

```python
df=df[df['Marks']>=75]
df=df.drop(['Age'],axis=1)
print("\nFiltering and Dropping func:\n",df)
```
